We will play around with English sentences on spaCy tool for extracting POS tags and dependency trees

````python
import spacy
from spacy import displacy
````

````python
nlp = spacy.load("en_core_web_sm")
````

````python
doc = nlp("I know my worth")
displacy.render(doc, style="dep", jupyter=True)
````

## The syntax tree is:
```
      know (ROOT / main verb)
     /    \
   I       worth
 (nsubj)    |
           my
          (poss)
```
## POS tags are:
PRON (PRP) — personal pronoun: I
Stands in for a person: here, subject form.

VERB (VBP) — verb, non-3rd-person singular present: know
Fine tag VBP = present (I know): coarse VERB just says it’s a verb.

PRON (PRP$) — possessive pronoun: my
Marks ownership/relation (“whose?”).

NOUN (NN) — common noun, singular: worth
The thing being referred to as the object.

## Dependencies in this sentence (arrows - this word modifies / depends on that word):
ROOT — know
Main predicate: the whole sentence hangs off this verb.

nsubj (nominal subject) — I → know
Who knows? I.

dobj (direct object) — worth → know
What is known? worth (the object of the verb).

poss (possession modifier) — my → worth
Whose worth? my — attached to the noun, not to the verb.

````python
doc = nlp("The quick brown fox jumped over the lazy dog.")
displacy.render(doc, style="dep", jupyter=True)
````

## The syntax tree is:
```
        jumped  (ROOT / main verb)
       /      \
   fox          over  (prep)
  / | \           |
The quick brown  dog  (pobj)
                 / \
              the lazy
```
## POS tags are:
DET (DT) — determiner: The, the
Marks which/referring info about a noun (fox/dog).

ADJ (JJ) — adjective: quick, brown, lazy
Describes a noun.

NOUN (NN) — common noun, singular: fox, dog
The main entity words.

VERB (VBD) — verb, past tense: jumped
Fine tag VBD = past; coarse VERB just says it’s a verb.

ADP (IN) — adposition (here a preposition): over
Introduces a relation like location/direction.

## Dependencies in this sentence (arrows - this word modifies / depends on that word):
ROOT — jumped
Main predicate; the whole sentence hangs off this verb.

nsubj (nominal subject) — fox → jumped
Who did the jumping? The fox.

amod (adjectival modifier) — quick/brown → fox, lazy → dog
Adjectives describing the noun.

det (determiner) — The → fox, the → dog
Article attached to its noun.

prep (prepositional modifier) — over → jumped
Prepositional phrase hanging off the verb (“jumped over …”).

pobj (object of preposition) — dog → over
Noun inside the PP: over the lazy dog.
