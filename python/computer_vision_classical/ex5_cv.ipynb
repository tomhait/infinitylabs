We will use histogram qualization to understand what is in the mystery picture

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = cv2.imread('mystery.png')
rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
print("Shape:", image.shape)
print("Min / max pixel:", gray.min(), gray.max())
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
axes[0].imshow(rgb)
axes[0].set_title("Original mystery.png")
axes[0].axis("off")
axes[1].hist(gray.ravel(), bins=256, range=(0, 255), color="gray")
axes[1].set_title("Histogram (original)")
axes[1].set_xlabel("Pixel intensity")
axes[1].set_ylabel("Count")
plt.tight_layout()
plt.show()
```

```python
equalized_gray = cv2.equalizeHist(gray)
fig, axes = plt.subplots(1, 3, figsize=(18, 6))
axes[0].imshow(rgb)
axes[0].set_title("Original")
axes[0].axis("off")
axes[1].imshow(equalized_gray, cmap="gray")
axes[1].set_title("Equalized (grayscale)")
axes[1].axis("off")
axes[2].hist(equalized_gray.ravel(), bins=256, range=(0, 255), color="steelblue")
axes[2].set_title("Histogram after equalization")
axes[2].set_xlabel("Pixel intensity")
plt.tight_layout()
plt.show()
```

equalizeHist evenly distributes pixel brightness across the entire spectrum (0–255) to bring out obscured details.
