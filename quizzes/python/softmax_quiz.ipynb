```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def softmax(x, T = 1.0):
    exp_x = np.exp((x - np.max(x))/T)
    return exp_x / np.sum(exp_x)
```

```python
x = np.arange(1, 1001)

T_values = [0.5, 1.0, 2.0, 3.0]

plt.figure(figsize=(12, 8))

for T in T_values:
    softmax_output = softmax(x, T=T)
    plt.plot(x, softmax_output, label=f'T = {T}')

plt.title('Softmax Output for Different T')
plt.xlabel('Input Value (x)')
plt.ylabel('Softmax Probability (log scale)')
plt.yscale('log')
plt.legend()
plt.grid(True)
plt.show()
```

T > 1.0 -> smoother probabilities - many classes get similar and low probs.
T < 1.0 -> Less spread, few classes get high prob.
