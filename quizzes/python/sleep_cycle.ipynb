We will model and cluster the sleep cycle in Israel

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
samp1 = np.random.normal(loc=3, scale=1, size=80000)
samp2 = np.random.normal(loc=14, scale=1, size=20000)

samples = np.concatenate((samp1, samp2))
```

```python
plt.figure(figsize=(10, 6))
plt.hist(samples, bins=50, density=True, alpha=0.7, color='skyblue', edgecolor='black')
plt.title('Histogram of Sleep Cycle In Israel')
plt.xlabel('Hour')
plt.ylabel('Density')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python
X = samples.reshape(-1, 1)

kmeans = KMeans(n_clusters=2, random_state=42, n_init=10)
kmeans.fit(X)

cluster_labels = kmeans.labels_
cluster_centers = kmeans.cluster_centers_

print("Cluster Centers:", cluster_centers.flatten())
```

As expected, K-means with K=2 resulted with two cluster centers at 2.99 (night sleep) and 13.99 (siesta) matching to the mean of the normal distributions.
